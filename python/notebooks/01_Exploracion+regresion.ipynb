# Regresión

```python
from sklearn.model_selection import train_test_split
```

```python
from sklearn.linear_model import LinearRegression
```

```python
df = data_frame[['Petal.Length', 'Petal.Width']]
df
```

```python
train_set, test_set = train_test_split(df, test_size=0.2, random_state=42)
```

```python
df_copy = train_set.copy()
df_copy.corr()
```

```python
import matplotlib.pyplot as plt
```

```python
plt.scatter(df_copy['Petal.Length'], df_copy['Petal.Width'])
```

```python
train_set_x = df_copy.drop(['Petal.Width'], axis=1)
train_set_label = df_copy['Petal.Width']
```

```python
lin_reg = LinearRegression()
lin_reg.fit(train_set_x,train_set_label)
```

```python
print("Coeficiente: ", lin_reg.coef_)
print("Intercept: ", lin_reg.intercept_)
```

```python
import numpy as np
valor = pd.DataFrame({'Petal.Length': [4]})  # Crear un DataFrame con la columna esperada
prediccion = lin_reg.predict(valor)
prediccion
```
